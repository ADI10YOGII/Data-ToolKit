```python
# merging the two dataframes

# see notes
# merge based on common column, inner merge(intersection)
# merge based on left dataframe, left merge(All the elements of dataframe A will be their in the merge)
# merge based on right dataframe,right merge(All the elements of datafrane B will be their in the merge)

# outer merge(All the elements of the dataframes )
# cross merge(Every permutation of rows are there)
```

```python
df1=pd.DataFrame({'key1': [1,2,4,5,6],'key2':[4,5,6,7,8],'key3':[3,4,5,6,6]})
```

```python
df1
```

```python
df2=pd.DataFrame({'key1': [1,2,45,6,67],'key4':[56,5,6,7,8],'key5':[3,56,5,6,6]})
```

```python
df2
```

```python
pd.merge(df1,df2,how='inner')
```

```python
pd.merge(df1,df2,how='left')
```

```python
pd.merge(df1,df2,how='right')
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,how='outer') # all the common elements of the common column from both the dataframe
```

```python
pd.merge(df1,df2,how='cross') # just like cartesian
```

```python
df1
```

```python
df2
```

```python
# merging when column name are not same
```

```python
pd.merge(df1,df2,how='left',left_on='key2',right_on='key4')
```
